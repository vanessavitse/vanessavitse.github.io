Now, the Mordell-Weil theorem, precised by Mazur, states that the group of $\mathbb Q$-rational points of an elliptic curve defined over $\mathbb Q$ is isomorphic to $H \times \mathbb Z^r$, where $H$ is finite (the torsion subgroup) of cardinality $\leq 12$, and $r$ is an integer called the rank of the elliptic curve.

Our strategy for solving $(*)$ will consist in computing the successive multiples of a non-torsion point on $\mathcal C$, until one is found with only positive coordinates. But before that, we need to put $\mathcal C$ in Weierstrass form.

(In homogeneous coordinates, a projective transformation consists simply in multiplying the coordinates by an invertible matrix $A$.)

## II. The elliptic curve factorization method



The elliptic curve factorization method (ECM) was introduced by H. Lenstra in 1987, as a refinement of Pollard's $p-1$ algorithm. It is currently the best algorithm for finding "small" factors (up to $\approx$ 50 digits) of large integers; for products of larger prime, the number field sieve is more efficient. The idea behind this method is to work with elliptic curves defined over $\mathbb Z/N\mathbb Z$ (which is not a field!), where $N$ is the integer to be factorized. 



### A. Elliptic curves mod N

Let $N$ be an odd composite integer. We say that a triple $(x,y,z) \in (\mathbb Z/N\mathbb Z)^3$ is *primitive* if the gcd of $x,y,z$ and $N$ is equal to 1. For instance, in $\mathbb Z/15 \mathbb Z$, the triples $(3,5,0)$ and $(2,4,6)$ are primitive, but $(3,6,9)$ is not.

We define the **projective plane** over $\mathbb Z/N\mathbb Z$ as $\quad \mathbb P^2(\mathbb Z/N\mathbb Z) = \{$primitive triples $(x,y,z) \in (\mathbb Z/N\mathbb Z)^3\} / \sim$ \
where $(x,y,z) \sim (ux,uy,uz)$ for any $u \in (\mathbb Z/N\mathbb Z)^\times$. 


It is straightforward to see that for any divisor $d$ of $N$ there is a well-defined reduction map 
$$\begin{array}{rcl} \mathbb P^2(\mathbb Z/N\mathbb Z) & \to & \mathbb P^2(\mathbb Z/d\mathbb Z) \\ [x:y:z] & \mapsto & [x \!\!\mod d : y \!\!\mod d : z \!\!\mod d]\end{array}$$
Furthermore, if $N = \prod_i p_i^{\alpha_i}$, then by Chinese remaindering we can see that $\ \displaystyle \mathbb P^2(\mathbb Z/N\mathbb Z) \simeq \prod_i \mathbb P^2(\mathbb Z/p_i^{\alpha_i}\mathbb Z)$

For $a,b$ such that $4a^3+27b^2 \in (\mathbb Z/N\mathbb Z)^\times$, the **elliptic curve** $E$ over $\mathbb Z/N\mathbb Z$ of homogeneous Weierstrass equation $Y^2Z=X^3+aXZ^2+bZ^3$ is simply the set of solutions of this equation in $\mathbb P^2(\mathbb Z/N\mathbb Z)$. For any prime divisor $p$ of $N$, we have again a well-defined reduction map 
$$\pi_p : \begin{array}{rcl} E & \to & E_p \\ [x:y:z] & \mapsto & [x \!\!\mod p : y \!\!\mod p : z \!\!\mod p]\end{array}$$
where $E_p$ is the elliptic curve of equation $Y^2Z=X^3+aXZ^2+bZ^3$ defined over $\mathbb Z/p\mathbb Z$, which this time is a field.

The usual addition formulae, in homogeneous coordinates, still work over $\mathbb Z/N\mathbb Z$ and allow to define a group law $+$ on $E$, such that the above reduction-mod-$p$ map $\pi_p$ is a group morphism for all prime divisors $p$ of $N$.

### B. Smooth integers and the ECM algorithm


Let $B \in \mathbb N^*$. We say that an integer $m$ is **$B$-powersmooth** if for all prime powers $p^\alpha$ dividing $m$, the inequality $p^\alpha \leq B$ holds. In other words, $m$ is divisible by $\displaystyle e_B = \prod_{p\ \mathrm{prime},\ p\leq B} p^{\lfloor \log(B)/\log(p)\rfloor}$.


4. Let $E$ be an elliptic curve over $\mathbb Z/N\mathbb Z$ together with a random point $P\in E$. Assume that $N$ has a prime divisor $p$ such that card($\pi_p(E)$) is $B$-powersmooth. \
Explain how the computation of $e_B P$ on $E$ may give a factor of $N$.
5. Use this idea to factorize $N=188099$ with $B=100$ (relevant SageMath commands: `n.is_prime()` and `next_prime(n)`).

Of course, if a given elliptic curve $E$ is defined modulo $N$, then in general card($\pi_p(E)$) will not be $B$-powersmooth for any divisor $p$ of $N$ and reasonably small $B$. However, we have plenty of choices for $E$. So these are the heuristics behind ECM factorization:

- If $N$ has a relatively small prime factor $p$, then around $p$ there is a non-negligible proportion of numbers that are $B$-powersmooth, for a small $B$. 
- If $E$ is a random elliptic curve over $\mathbb Z/N\mathbb Z$, then card($\pi_p(E)$) is a random integer around $p$.
- Thus there is a non-negligible proportion of curves $E$ for which card($\pi_p(E)$) is $B$-powersmooth. By picking sufficiently many random curves, we will ultimately find such a curve, allowing to find $p$ by computing $e_B P$ for a random point $P\in E$. 

These heurisitics can be made more rigorous; nevertheless, finding the optimal value of $B$ is far from obvious (and depends on the expected size of the smallest factor of $N$).  

6. Implement these idea and test your program. You can try to find factors of the first non-prime Fermat numbers $2^{2^n}+1$ for $n\geq 5$ (check the relevant wikipedia page to understand why the values $n=7$ or $8$ are much harder than the other values up to 12).
